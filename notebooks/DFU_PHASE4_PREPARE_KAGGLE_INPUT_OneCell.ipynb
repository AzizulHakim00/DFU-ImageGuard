```python
# DFU Phase-4: export frozen inputs from Google Drive to private Kaggle Dataset
import urllib.request, hashlib

VERSION = "DFU_PHASE4_PREPARE_KAGGLE_INPUT_ONECELL_V1_20260812"
SOURCE_COMMIT = "12df25b74c152caf9810d3ed6e8c2eb317ab02c6"
SOURCE_PATH = "scripts/phase4_kaggle_input_exporter_v1.py"
EXPECTED_GIT_BLOB = "5ece3193a5af5ab0f78c8fb926b25343d1797e16"
URL = f"https://raw.githubusercontent.com/AzizulHakim00/DFU-ImageGuard/{SOURCE_COMMIT}/{SOURCE_PATH}"

def git_blob_sha(raw):
    return hashlib.sha1(b"blob " + str(len(raw)).encode() + b"\0" + raw).hexdigest()

print("="*100)
print(VERSION)
print("READ-ONLY FROZEN-INPUT EXPORTER | NO TRAINING | NO EXTERNAL DATASET DOWNLOAD")
print("="*100)
raw = urllib.request.urlopen(URL, timeout=120).read()
actual = git_blob_sha(raw)
if actual != EXPECTED_GIT_BLOB:
    raise RuntimeError(f"Exporter source Git-blob mismatch: expected={EXPECTED_GIT_BLOB} actual={actual}")
print("Exporter source Git blob: PASS", actual)
text = raw.decode("utf-8")
compile(text, "phase4_kaggle_input_exporter_v1.py", "exec")
print("Exporter source compile: PASS")
exec(compile(text, "phase4_kaggle_input_exporter_v1.py", "exec"), globals())
```
