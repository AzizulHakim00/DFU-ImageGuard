Removes the false direct-source label_name marker check before running the pinned plain-Python Repair-7 runner.

```python
import urllib.request

VERSION = 'V5_FALSE_LABEL_MARKER_FIXED_20260809_2325'
URL = 'https://raw.githubusercontent.com/AzizulHakim00/DFU-ImageGuard/ffe4ebc3006f7c25eec413caf09f1dd2b66f510b/notebooks/repair7_local_upload/dfu_repair7_plain_py_cpu_v3.py'

print(VERSION)
raw = urllib.request.urlopen(URL, timeout=120).read()
code = raw.decode('utf-8')

bad = "    '[\"image_id\", \"group_id\", \"label\", \"label_name\", \"relative_path\"]',\n"
count = code.count(bad)
if count != 1:
    raise RuntimeError(f'False label_name marker patch expected once, found {count}')
code = code.replace(bad, '', 1)

if "'[\"image_id\", \"group_id\", \"label\", \"label_name\", \"relative_path\"]'" in code:
    raise RuntimeError('False direct-source label_name marker still remains')
if 'data[\"label_name\"]=np.where' not in code:
    raise RuntimeError('Real label_name reconstruction safety check is missing')

compile(code, 'dfu_repair7_plain_py_cpu_v3_markerfixed.py', 'exec')
print('False direct-source label_name marker: REMOVED')
print('Real label_name reconstruction patch: PRESERVED')
print('Patched plain-Python runner compile: PASS')
print('Next: upload DFU_GOOD38_RECOVERY_EVIDENCE.zip when prompted')
exec(compile(code, 'dfu_repair7_plain_py_cpu_v3_markerfixed.py', 'exec'), globals())
```
