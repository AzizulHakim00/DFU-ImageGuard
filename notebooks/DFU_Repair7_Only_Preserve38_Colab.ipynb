# DFU Repair-7 Only — Preserve 38 Good Trials
Direct Colab launcher. Repairs only the seven incompatible fold-1 trials and preserves the 38 good trials.


```python
import base64, gzip, hashlib, json, urllib.request

PAYLOAD_URL = 'https://raw.githubusercontent.com/AzizulHakim00/DFU-ImageGuard/df218aea881a2ee416d81c34c93280222a147b99/notebooks/repair7_payload/DFU_Repair7_Only_Preserve38.ipynb.gz.b64'
EXPECTED_NOTEBOOK_SHA256 = '368e13cd015137563875f5f274e2cd594f47384e0fee9b656445e814d60a238a'

encoded = urllib.request.urlopen(PAYLOAD_URL, timeout=120).read().strip()
notebook_bytes = gzip.decompress(base64.b64decode(encoded, validate=True))
actual_sha = hashlib.sha256(notebook_bytes).hexdigest()
if actual_sha != EXPECTED_NOTEBOOK_SHA256:
    raise RuntimeError(f'Repair-7 payload SHA mismatch: {actual_sha} != {EXPECTED_NOTEBOOK_SHA256}')
print('Pinned Repair-7 payload verification: PASS')

nb = json.loads(notebook_bytes.decode('utf-8'))
code_cells = [c for c in nb['cells'] if c.get('cell_type') == 'code']
if len(code_cells) != 1:
    raise RuntimeError(f'Expected exactly one executable repair cell, found {len(code_cells)}')
repair_code = ''.join(code_cells[0]['source'])
exec(compile(repair_code, 'DFU_Repair7_Only_Preserve38.py', 'exec'), globals())
```
