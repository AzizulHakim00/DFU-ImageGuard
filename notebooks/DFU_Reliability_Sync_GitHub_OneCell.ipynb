# DFU Reliability GitHub Sync — no retraining
Pushes a fully verified saved run to the reliability-results branch. Full .pt files are split into SHA-256-verified chunks; portable FP16 checkpoints are also copied normally. Training never depends on this sync step.

```python
import os, sys, json, shutil, subprocess
from pathlib import Path
subprocess.run([
    sys.executable, '-m', 'pip', 'install', '-q',
    'timm>=1.0.9', 'kagglehub>=0.3', 'ImageHash>=4.3',
    'scikit-learn>=1.5', 'scipy>=1.13', 'matplotlib>=3.9',
    'pandas>=2.2', 'Pillow>=10.4', 'tabulate>=0.9'
], check=True)
REPO='https://github.com/AzizulHakim00/DFU-ImageGuard.git'
BRANCH='reliability-framework-v1'
WORK=Path('/content/DFU-ImageGuard-reliability')
if WORK.exists(): shutil.rmtree(WORK)
subprocess.run(['git','clone','--depth','1','--branch',BRANCH,REPO,str(WORK)],check=True)
os.chdir(WORK)
sys.path.insert(0,str(WORK))
for name in list(sys.modules):
    if name=='src' or name.startswith('src.'):
        del sys.modules[name]

from src.reliability_launcher import mount_drive_verified
from src.reliability_sync import sync_reliability_run_to_github
project_root=mount_drive_verified()
run_root=project_root/'runs'/'DFU_RELIABILITY_FINAL_V1'
result=sync_reliability_run_to_github(run_root=run_root)
print(json.dumps(result,indent=2,default=str))
```
