# DFU Repair-7 — Tiny Loader
Small Colab-safe loader for the pinned Local Upload Direct CPU v2 repair runner.

```python
import json, urllib.request

PINNED_NOTEBOOK = "https://raw.githubusercontent.com/AzizulHakim00/DFU-ImageGuard/eba940a5c7754fed02401af1296251953e281041/notebooks/DFU_Repair7_LOCAL_UPLOAD_DIRECT_CPU_V2_GITHUB.ipynb"

raw = urllib.request.urlopen(PINNED_NOTEBOOK, timeout=120).read()
nb = json.loads(raw.decode("utf-8"))
cells = [c for c in nb.get("cells", []) if c.get("cell_type") == "code"]
if len(cells) != 1:
    raise RuntimeError(f"Expected exactly one pinned code cell, found {len(cells)}")

code = "".join(cells[0]["source"])
compile(code, "DFU_Repair7_LOCAL_UPLOAD_DIRECT_CPU_V2.py", "exec")

print("DFU Repair-7 tiny loader: PASS")
print("Pinned CPU v2 code: fetched and compiled")
print("Next: upload DFU_GOOD38_RECOVERY_EVIDENCE.zip when prompted")

exec(compile(code, "DFU_Repair7_LOCAL_UPLOAD_DIRECT_CPU_V2.py", "exec"), globals())
```
