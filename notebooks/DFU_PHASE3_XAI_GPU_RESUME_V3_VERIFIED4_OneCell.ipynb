```python
# DFU Phase-3 XAI — GPU Resume-Safe V3 VERIFIED4 — one cell
import urllib.request, base64, zlib, hashlib

VERSION = "DFU_PHASE3_XAI_GPU_RESUME_V3_VERIFIED4_20260812"
PAYLOAD_COMMIT = "5118579d0c9f83b547a08b76b97355be1d6e2dbf"
EXPECTED_SOURCE_SHA256 = "9ba6e35e8e19c771fc7c6e8fbfcd27b7c2c882ff16878206f1e3660ffaa6057b"
PARTS = ['scripts/gpu_xai_resume_v3_verified4/part_00.txt', 'scripts/gpu_xai_resume_v3_verified4/part_01.txt', 'scripts/gpu_xai_resume_v3_verified4/part_02.txt', 'scripts/gpu_xai_resume_v3_verified4/part_03.txt']
BASE = "https://raw.githubusercontent.com/AzizulHakim00/DFU-ImageGuard/" + PAYLOAD_COMMIT + "/"

print("="*100)
print(VERSION)
print("VALIDATED LOADER | GPU-AWARE | READ-ONLY | RESUME-SAFE")
print("Primary training: FORBIDDEN")
print("Primary 45-trial modification: FORBIDDEN")
print("All important Phase-2 and Phase-3 tables/plots display in THIS SAME OUTPUT CELL.")
print("="*100)

chunks=[]
for path in PARTS:
    txt=urllib.request.urlopen(BASE+path, timeout=120).read().decode("ascii").strip()
    print(f"Fetched {path}: {len(txt):,} chars")
    chunks.append(txt)
payload="".join(chunks)
try:
    compressed=base64.b64decode(payload, validate=True)
    source=zlib.decompress(compressed)
except Exception as exc:
    raise RuntimeError(f"Verified V3 payload reconstruction failed: {type(exc).__name__}: {exc}") from exc
actual=hashlib.sha256(source).hexdigest()
if actual != EXPECTED_SOURCE_SHA256:
    raise RuntimeError(f"Decoded V3 source SHA256 mismatch: expected={EXPECTED_SOURCE_SHA256} actual={actual}")
print("Decoded V3 source SHA256: PASS", actual)
text=source.decode("utf-8")
compile(text, "dfu_phase3_xai_gpu_resume_v3.py", "exec")
print("Decoded V3 source compile: PASS")
print("Starting GPU-aware resume-safe XAI...")
exec(compile(text, "dfu_phase3_xai_gpu_resume_v3.py", "exec"), globals())
```

```python
from google.colab import drive
drive.mount("/content/drive", force_remount=False)

from pathlib import Path
import json
import pandas as pd
from IPython.display import display, Image, Markdown

RUN_ROOT = Path(
    "/content/drive/MyDrive/DFU-ImageGuard/runs/"
    "RELIABLE_DFU_CV_V3_MISSING38"
)

P3 = RUN_ROOT / "PHASE3_XAI_FULL"

print("=" * 100)
print("DFU PHASE-3 XAI SAVED RESULT CHECK")
print("=" * 100)
print("Checking:", P3)

if not P3.exists():
    raise FileNotFoundError(
        "PHASE3_XAI_FULL folder not found:\n" + str(P3)
    )

print("PHASE3_XAI_FULL: FOUND")

# ------------------------------------------------------------
# 1. FIND ALL IMPORTANT FILES
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("IMPORTANT FILES")
print("=" * 100)

patterns = [
    "*PROGRESS*.json",
    "*VERIFICATION*.json",
    "*SUMMARY*.json",
    "*FINAL_EXPORT*.zip",
    "*.pkl",
]

found_important = []

for pattern in patterns:
    for p in sorted(P3.rglob(pattern)):
        if p not in found_important:
            found_important.append(p)

for p in found_important:
    print(
        "FOUND |",
        str(p.relative_to(P3)),
        "|",
        round(p.stat().st_size / (1024 ** 2), 3),
        "MB"
    )

if not found_important:
    print("No important final/progress files found.")

# ------------------------------------------------------------
# 2. READ PROGRESS FILES
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("PROGRESS")
print("=" * 100)

progress_files = sorted(P3.rglob("*PROGRESS*.json"))

if progress_files:
    for p in progress_files:
        print("\nFILE:", p.relative_to(P3))
        try:
            obj = json.loads(p.read_text(encoding="utf-8"))
            print(json.dumps(obj, indent=2, default=str))
        except Exception as e:
            print("READ ERROR:", repr(e))
else:
    print("No progress JSON found.")

# ------------------------------------------------------------
# 3. READ VERIFICATION FILES
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("VERIFICATION")
print("=" * 100)

verification_files = sorted(P3.rglob("*VERIFICATION*.json"))

verification_pass = False

if verification_files:
    for p in verification_files:
        print("\nFILE:", p.relative_to(P3))

        try:
            obj = json.loads(p.read_text(encoding="utf-8"))
            print(json.dumps(obj, indent=2, default=str))

            txt = json.dumps(obj).upper()

            if (
                '"STATUS": "PASS"' in txt
                or '"VERIFICATION": "PASS"' in txt
                or '"PASS"' in txt
            ):
                verification_pass = True

        except Exception as e:
            print("READ ERROR:", repr(e))
else:
    print("No verification JSON found.")

# ------------------------------------------------------------
# 4. CSV TABLE INVENTORY
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("SAVED TABLES")
print("=" * 100)

csv_files = sorted(P3.rglob("*.csv"))

print("CSV count:", len(csv_files))

table_inventory = []

for p in csv_files:
    try:
        df = pd.read_csv(p)

        table_inventory.append({
            "file": str(p.relative_to(P3)),
            "rows": len(df),
            "columns": len(df.columns)
        })

    except Exception as e:
        table_inventory.append({
            "file": str(p.relative_to(P3)),
            "rows": "ERROR",
            "columns": str(e)
        })

if table_inventory:
    display(pd.DataFrame(table_inventory))

# ------------------------------------------------------------
# 5. XAI STATUS TABLE
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("XAI METHOD STATUS")
print("=" * 100)

status_files = [
    p for p in csv_files
    if "status" in p.name.lower()
    and "xai" in p.name.lower()
]

if status_files:
    for p in status_files:
        print("\nSTATUS FILE:", p.relative_to(P3))

        df = pd.read_csv(p)
        display(df)

        status_cols = [
            c for c in df.columns
            if "status" in c.lower()
        ]

        for c in status_cols:
            print("\nStatus counts using:", c)
            print(df[c].value_counts(dropna=False).to_string())

else:
    print("No XAI status CSV found.")

# ------------------------------------------------------------
# 6. XAI MAP COUNT
# ------------------------------------------------------------

npy_files = sorted(P3.rglob("*.npy"))
png_files = sorted(P3.rglob("*.png"))
pdf_files = sorted(P3.rglob("*.pdf"))
json_files = sorted(P3.rglob("*.json"))
zip_files = sorted(P3.rglob("*.zip"))

print("\n" + "=" * 100)
print("ARTIFACT COUNTS")
print("=" * 100)

print("NPY attribution maps :", len(npy_files))
print("PNG figures          :", len(png_files))
print("PDF figures          :", len(pdf_files))
print("CSV tables           :", len(csv_files))
print("JSON files           :", len(json_files))
print("ZIP files            :", len(zip_files))

# ------------------------------------------------------------
# 7. SHOW LATEST FIGURES IN OUTPUT CELL
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("LATEST SAVED XAI FIGURES")
print("=" * 100)

latest_png = sorted(
    png_files,
    key=lambda p: p.stat().st_mtime,
    reverse=True
)[:10]

if latest_png:
    for p in latest_png:
        print("\nFIGURE:", p.relative_to(P3))
        display(Image(filename=str(p), width=1000))
else:
    print("No PNG figures found.")

# ------------------------------------------------------------
# 8. CHECK FINAL EXPORT
# ------------------------------------------------------------

final_zip = list(P3.rglob("PHASE3_XAI_FINAL_EXPORT.zip"))

print("\n" + "=" * 100)
print("FINAL VERDICT")
print("=" * 100)

if verification_pass and final_zip:
    print("PHASE-3 XAI STATUS: COMPLETE")
    print("Verification PASS found.")
    print("Final export found:")
    print(final_zip[0])

elif final_zip:
    print("PHASE-3 XAI STATUS: FINAL ZIP EXISTS")
    print("But PASS verification was not automatically confirmed.")
    print("Check the verification JSON printed above.")

elif len(npy_files) > 0 or len(csv_files) > 0:
    print("PHASE-3 XAI STATUS: PARTIAL RESULTS SAVED")
    print("The web refresh did NOT remove completed saved work.")
    print("Resume-safe XAI notebook can continue from saved maps.")

else:
    print("PHASE-3 XAI STATUS: NO SAVED PHASE-3 RESULTS FOUND")

print("=" * 100)
```
