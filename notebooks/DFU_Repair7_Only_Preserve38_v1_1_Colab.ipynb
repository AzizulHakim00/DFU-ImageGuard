# DFU Repair-7 Only v1.1 — Robust Drive Mount
Repairs only the seven incompatible fold-1 trials; 38 good trials remain protected.


```python
import base64, gzip, hashlib, json, urllib.request

PAYLOAD_URL = 'https://raw.githubusercontent.com/AzizulHakim00/DFU-ImageGuard/00694715d022b9ea6ce32532328491c5284d8523/notebooks/repair7_payload/DFU_Repair7_Only_Preserve38_v1_1.ipynb.gz.b64'
EXPECTED_NOTEBOOK_SHA256 = '879e40ac41e27ae473486386048e52e9d683e95409aa542a6a9526d219133a99'

encoded = urllib.request.urlopen(PAYLOAD_URL, timeout=120).read().strip()
notebook_bytes = gzip.decompress(base64.b64decode(encoded, validate=True))
actual_sha = hashlib.sha256(notebook_bytes).hexdigest()
if actual_sha != EXPECTED_NOTEBOOK_SHA256:
    raise RuntimeError(f"Repair-7 v1.1 payload SHA mismatch: {actual_sha} != {EXPECTED_NOTEBOOK_SHA256}")
print("Pinned Repair-7 v1.1 payload verification: PASS")

nb = json.loads(notebook_bytes.decode("utf-8"))
code_cells = [c for c in nb["cells"] if c.get("cell_type") == "code"]
if len(code_cells) != 1:
    raise RuntimeError(f"Expected exactly one executable repair cell, found {len(code_cells)}")
repair_code = "".join(code_cells[0]["source"])
exec(compile(repair_code, "DFU_Repair7_Only_Preserve38_v1_1.py", "exec"), globals())
```
